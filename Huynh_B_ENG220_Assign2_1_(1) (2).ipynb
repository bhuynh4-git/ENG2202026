#1 Multiplication Table

```python
number = int(input("Enter a number: "))

for i in range(10):
  print(number * (i+1))
```

2. Operations

```python
def add(num_1, num_2):
  return num_1 + num_2

def subtract(num_1, num_2):
  return num_1 - num_2

def multiply(num_1, num_2):
  return num_1 * num_2

def divide(num_1, num_2):
  if num_2 == 0:
    return "Division by zero is not possible"
  else:
    return num_1 / num_2


num_1 = int(input("Enter your 1st number: "))
num_2 = int(input("Enter your 2nd number: "))
operation = input("Enter your operation: ")


if operation == "+":
  print(add(num_1, num_2))
elif operation == "-":
  print(subtract(num_1, num_2))
elif operation == "*":
  print(multiply(num_1, num_2))
elif operation == "/":
  print(divide(num_1, num_2))
else:
  print("Operation input was invalid, only use symbols of + - * /")
```
